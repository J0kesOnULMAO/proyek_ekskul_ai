
### Preprocessing Data


```python
import pandas as pd

data = {
"nama": ["Andi", "Budi", "Citra", "Dini", "Eka", "Fajar", "Fajar"],
"jam_belajar": [5, 2, 8, None, 6, 3, 3],
"jumlah_latihan": [10, 3, 15, 2, None, 5, 5],
"kehadiran": [90, 60, 95, 50, 85, 70, 70],
"gender": ["Laki-laki", "Laki-laki", "Perempuan", "Perempuan",
"Perempuan", "Laki-laki", "Laki-laki"],
"lulus": ["Ya", "Tidak", "Ya", "Tidak", "Ya", "Tidak", "Tidak"]
}
df = pd.DataFrame(data)

print(df)
```

#### Check missing value


```python
# cek missing value

df.isna().sum()
```

```python
#mengisi nilai kosong
ratajam = df["jam_belajar"].mean()
df['jam_belajar'] = df['jam_belajar'].fillna(ratajam)

ratalat = df["jumlah_latihan"].mean()
df['jumlah_latihan'] = df['jumlah_latihan'].fillna(ratalat).round(1)

print(df)
```

```python
#cek dulplikat
df.duplicated().sum()
df = df.drop_duplicates()
print (df)
```

```python
#cek tipe data
df.info()
df["jumlah_latihan"] = df["jumlah_latihan"].astype(int)
print(df)
```

```python
df["lulus_encode"] = df["lulus"].map({
    "Ya" : 1,
    "Tidak" : 0
})
print(df)
```

```python
gender_encode = pd.get_dummies(df["gender"])
print(gender_encode)
```

```python
#scale data numerik

df["jam_belajar_scaled"] = (
(df["jam_belajar"]- df["jam_belajar"].min()) /
(df["jam_belajar"].max() -df["jam_belajar"].min())
)

df["jam_belajar_scaled"] = df["jam_belajar_scaled"].round(1)
print(df[["jam_belajar", "jam_belajar_scaled"]])
```

```python
# data outliner

df.describe()
```
